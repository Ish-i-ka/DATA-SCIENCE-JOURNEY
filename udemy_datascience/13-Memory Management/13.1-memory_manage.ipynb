# Reference Counting
It is a primary method Python uses to manage memory. Each object maintains a count of the no. of references pointing to it. When the count drops to 0, the memory occupied by the object gets deallocated


```python
import sys 

a = []
print(sys.getrefcount(a)) 
```

```python
del a
print(sys.getrefcount(a)) #error bcoz a is already deleted and its reference count is not available count = 0
```

# Garbage Collection
Python uses a cyclic garbage collector to handle reference cycles. Reference cycles occur when objects reference each other, preventing their reference counts from reaching 0.


```python
import gc

#enable garbage collector
gc.enable()
```

```python
gc.disable()
```

```python
gc.collect() #collects all the garbage, basically gives the number of var which are not in use
```

```python
#Get Garbage collection stats
print(gc.get_stats())
```

```python
#get the unreachable objects
print(gc.garbage)
```

```python
import gc 

class MyObject:
    def __init__(self,name):
        self.name = name
        print(f"Object {self.name} created")
        
    def __del__(self):
        print(f"Object {self.name} deleted")
    
#object creation
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")

#Circular reference
obj1.ref = obj2
obj2.ref = obj1

del obj1
del obj2
#Now as both objects are deleted so they are garbage
#But they are not collected by garbage collector as they have circular reference
#So we need to manually trigger garbage collector
gc.collect()
        
```

```python
gc.get_stats()
```

```python
#Print the collected garbage
print(gc.garbage)
```

```python
#Generators for Memory efficiency

def generate_numbers(n):
    for i in range(n):
        yield i

#calling the generator
for num in generate_numbers(1000):
    print(num)
    if(num>9):
        break
```

```python
#Profiling Memory Usage with tracemalloc

import tracemalloc
def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()
    
    create_list()
    
    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')
    
    print("[ Top 100]")
    for stat in top_stats[:100]:
        print(stat)
```

```python
main()
```
