```python
#Handling missing values

import pandas as pd
df = pd.read_csv('sales.csv')
df.head()
```

```python
df.isnull()
```

```python
df.isnull().any(axis=1)     #if there is any null value or missing value then True will be shown against that row no. and axis=1 basically takes all columns together
```

```python
df.isnull().sum()   #how many total null values are there in each col 
```

```python
# fills the missing values with 0

df_filled = df.fillna(0)
df_filled.head(3)
```

```python
#filling missing values with mean of the column
df['Qty_fillNA'] = df['Quantity'].fillna(df['Quantity'].mean())
df
```

```python
df.dtypes
```

```python
df = df.rename(columns = {'Quantity':'Net Qty'})
df.head()
```

```python
#Change data type

df['Profit_new'] = df['Profit'].astype(float)
df.head()
```

```python
## if we want to apply any func to all the values of a col

df['New Qty']=df['Net Qty'].apply(lambda x:x**2)
df.head()
```

```python
## Data aggregating and grouping

#grouping category on the basis of mean of their net qty sold
grouped_mean = df.groupby('Category')['Net Qty'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Category','Sub-Category'])['Net Qty'].sum()
print(grouped_sum)
```

```python
df.groupby(['Category','Sub-Category'])['Net Qty'].mean()
```

```python
## grouping multiple cols with multiple aggregate functions

grouped_agg = df.groupby(['Category','Sub-Category'])['Net Qty'].agg(['mean','sum','count'])
grouped_agg
```

```python
## Joining operations

data = [
    {'Key':'A','Value1':1},
    {'Key':'B','Value1':2},
    {'Key':'C','Value1':3}
]
df1 = pd.DataFrame(data)
df2 = pd.DataFrame({'Key':['A','B','D'], 'Value2':[4,5,6]})
```

```python
df1
```

```python
df2
```

```python
#inner join on key column

pd.merge(df1,df2,on='Key',how = 'inner')
```

```python
#outer join on key column

pd.merge(df1,df2,on='Key',how = 'outer')
```

```python
#left outer join on key column

pd.merge(df1,df2,on='Key',how = 'left')
```

```python
#right outer join on key column

pd.merge(df1,df2,on='Key',how = 'right')
```
