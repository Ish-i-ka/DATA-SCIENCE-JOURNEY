```python
import logging

#create logger for module1
logger1 = logging.getLogger("module1")      #giving a name to the logger
logger1.setLevel(logging.DEBUG)

#create logger for module2
logger2 = logging.getLogger("module2")
logger2.setLevel(logging.WARNING)

#Configure logging settings
logging.basicConfig(          
    level=logging.DEBUG,
    format='%(asctime)s - %(name)s - %(levelname)s - %(message)s',
    datefmt='%Y-%m-%d %H:%M:%S'
)
```

```python
#log messages with different loggers

#by creating multiple loggers the messages are displayed along with respective modules and not only root

logger1.debug("This is a debug message of module1")
logger2.warning("This is a warning meesage of module2")
logger2.error("This is an error message")
```
