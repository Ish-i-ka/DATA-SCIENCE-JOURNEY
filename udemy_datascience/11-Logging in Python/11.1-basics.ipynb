```python
#Logging is dn to keep a track/logs of errors or events occuring while making any project 

import logging
#Configure basic logging settings
logging.basicConfig(level=logging.DEBUG)

#log messages
logging.debug("This is a debug msg")
logging.info("This is a info msg")
logging.warning("This is a warning msg")
logging.error("This is a error msg")
logging.critical("This is a critical msg")
```

### LOG LEVELS
1.DEBUG:Detailed information about cases dealing with problems
2.INFO:Confirmation of things are working as expected
3.WARNING:Indication of something unexpected happened or will happen but the software is working as expected.
4.ERROR:Due tp serious issue the software cant function properly
5.CRITICAL:A vry serious error indicating that the software itself stops running. 

```python
#configure logging
import logging

logging.basicConfig(
    filename='app.log',     #the log messages are stored into this file
    filemode='w',           #the file is opened in write mode
    level=logging.DEBUG,
    format='%(asctime)s-%(name)s-%(levelname)s-%(message)s',
    datefmt='%Y/%m/%d %H:%M:%S'
)

#log messages
logging.debug("This is a debug msg")
logging.info("This is a info msg")
logging.warning("This is a warning msg")
logging.error("This is a error msg")
logging.critical("This is a critical msg")
```
