```python
import sqlite3
```

```python
#connecting to an sqlite database

connection = sqlite3.connect('example.db')
print(connection)
```

```python
#cursor() is used to perform crud operations in the database tables
cursor=connection.cursor()
```

```python
#table creation
cursor.execute(''' 
Create Table If Not Exists employee(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department Text
    )                             
''')
#commit changes to database
connection.commit()
```

```python
cursor.execute('Select *from employee')
```

```python
#insert values
cursor.execute('''
Insert into employee(id,name,age,department) 
values("42","Ishika Saha","20","Data Scientist")              
''')
```

```python
#Update values

cursor.execute('''
Update employee
Set department='Finance'
Where id=42               
''')
```

```python
#Query data from table
cursor.execute('Select *from employee')
rows=cursor.fetchall()

#print queried data
for row in rows:
    print(row)
```

```python
#Delete data

cursor.execute('Delete from employee Where id="42"')
connection.commit()
```

```python
#Query data from table
cursor.execute('Select *from employee')
rows=cursor.fetchall()

#print queried data
for row in rows:
    print(row)
```
