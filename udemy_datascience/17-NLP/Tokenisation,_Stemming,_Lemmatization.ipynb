```python
corpus = """Hello everyone, I am Ms.Ishika Saha, a Final Year Student from the University of Engineering and Management. I hope you are doing good. I love to eat Pizzas! This is my start to NLTK.
"""
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
#Paragraphs --> Sentence using sent_tokenize

from nltk.tokenize import sent_tokenize
sent_tokenize(corpus)
```

```python
documents = sent_tokenize(corpus)
```

```python
type(documents)
```

```python
for sentence in documents:
  print(sentence)
```

```python
#Para --> words

from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
#Sentence --> words

for sentence in documents:
  print(word_tokenize(sentence))
```

```python
#for seperating the punctuations and treating them as seperate words

from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
#tree bank  word tokenizer
#any full stop in between the para or sentence is not treated as diff word,
#only the last ending full stop is treated as seperate word

from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

#**Stemming**

It is a process of reducing a word to its word stem or root which aree affixed to suffixes and prefixes and the root is known as lemma.

```python
words = ['eating','eats','eaten','write','writes','written','programs','programming','programmer','finally','finals']
```

```python
#PorterStemmer

from nltk.stem import PorterStemmer
stemming = PorterStemmer()
for word in words:
  print(word+"--> "+stemming.stem(word))
```

```python
#disadvantage of stemming -> it changes the meaning of word

stemming.stem('congratulations')
```

**RegexpStemmer Class**

NLTK provides this class that helps to implement Regular Expression Stemmer algorithms. It takes a single reg esp. nd removes prefix or suffix that matches the expression.

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('seats')

#s from end is removed only
```

**Snowball Stemmer**

```python
from nltk.stem import SnowballStemmer

snowball = SnowballStemmer('english')

words = ['eating','eats','eaten','write','writes','written','programs','programming','programmer','finally','finals']
for word in words:
  print(word+"--> "+snowball.stem(word))
```

```python
#lets compare porterStemmer nd SnowballStemmer

stemming.stem('sportingly'),stemming.stem('fairly')
```

```python
snowball.stem('sportingly'),snowball.stem('fairly')
```

Thus, Snowball Stemmer works better than PorterStemmer in some cases but both are not apt for building chatbots, for that we need lemmatization and not stemming.

**Lemmatization**

Its more powerful technique than stemming that gives lemma as output which is a root word rather than a root stem. This root word is a meaningful valid word.

**WordNetLemmatizer**

This class of NLTK is a thin wrapper around wordnet corpus. This class uses morphy() to the WordNet CorpusReader class to find a lemma.

```python
 nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

'''
POS: Noun-n, verb-v, adjective-a, adverb-r
'''
lemmatizer.lemmatize('going') #pos not mentioned means pos=n
```

```python
lemmatizer.lemmatize('going',pos='v')
```

```python
words = ['eating','eats','eaten','write','writes','written','programs','programming','programmer','finally','finals']

for word in words:
  print(word+"--> "+ lemmatizer.lemmatize(word,pos='v'))
```
