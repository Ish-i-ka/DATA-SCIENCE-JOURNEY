```python
import pandas as pd
import numpy as np
```

```python
#creating dataset
from sklearn.datasets import make_classification
X,y = make_classification(n_samples=1000,n_features=10,n_informative=3,n_classes=3,random_state = 60)
```

```python
y     #it shows 3 types of outputs
```

```python
#train test split

from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
#training
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(multi_class='ovr')
#ovr=One versus Rest
classifier.fit(X_train,y_train)
y_pred = classifier.predict(X_test)
```

```python
#Performance metrics

from sklearn.metrics import accuracy_score,classification_report,confusion_matrix
score = accuracy_score(y_test,y_pred)
print("Accuracy=",score)
print('Classification Report:')
print(classification_report(y_test,y_pred))
print('Confusion Matrix:')
print(confusion_matrix(y_test,y_pred))
```

##HyperParameter Tuning

1.Grid SearchCV

```python
model = LogisticRegression()
c_values = [100,10,1.0,0.1,0.01]
max_iters = [100, 200, 500, 1000] # Good to tune this for convergence
params =[
    {
        'multi_class': ['ovr'], # liblinear only works with 'ovr' for multiclass
        'penalty': ['l1', 'l2'], # liblinear supports L1 and L2
        'C': c_values,
        'solver': ['liblinear'],
        'max_iter': max_iters # Add max_iter
    },
    {
        'multi_class': ['ovr', 'multinomial'], # These solvers can handle both
        'penalty': ['l2'], # lbfgs, newton-cg, sag mainly support L2 (or none)
        'C': c_values,
        'solver': ['lbfgs', 'newton-cg', 'sag'],
        'max_iter': max_iters
    },
    {
        'multi_class': ['ovr', 'multinomial'], # saga can handle both
        'penalty': ['l1', 'l2', 'elasticnet'], # saga is versatile
        'C': c_values,
        'solver': ['saga'],
        'l1_ratio': [0.3, 0.5, 0.7], # Only used if penalty is 'elasticnet'
        'max_iter': max_iters
    }
]
```

```python
from sklearn.model_selection import GridSearchCV,StratifiedKFold
cv = StratifiedKFold()
grid = GridSearchCV(estimator=model,param_grid=params,cv=cv,scoring='accuracy')
grid.fit(X_train,y_train)
grid
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
y_pred_grid = grid.predict(X_test)      #pred output for grid searchCV
score = accuracy_score(y_test,y_pred_grid)
print("Accuracy=",score)
print('Classification Report:')
print(classification_report(y_test,y_pred_grid))
print('Confusion Matrix:')
print(confusion_matrix(y_test,y_pred_grid))
```

###Accuracy increased from 84% to 86%
