```python
import numpy as np
import pandas as ps
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons   #to create datasets that appear in the shape of a crescent
```

```python
X,y = make_moons(n_samples=250,noise=0.05)
```

```python
X,y
```

```python
#Visualizing the original dataset

plt.scatter(X[:,0],X[:,1])
```

```python
#Feature scaling

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.3)    #eps is the chosen epsilon value
```

```python
dbscan.fit(X_scaled)
```

```python
dbscan.labels_
```

```python
#Plotting the clustered data

plt.scatter(X[:,0],X[:,1],c=dbscan.labels_)
```
