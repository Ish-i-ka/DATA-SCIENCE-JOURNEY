```python
print(df.data)      #gives rows of values for all independent features
```

```python
print(df.target)      #gives values for dependent feature
```

```python
#Dataframe preparing

dataset = pd.DataFrame(df.data,columns=df.feature_names)
dataset
```

```python
#adding dependent feature
dataset['MedHouseVal'] = df.target
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
dataset.describe()
```

```python
dataset.corr()
```

```python
sns.heatmap(dataset.corr(),annot=True)
```

```python
#Independent and dependent features split

X = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]
```

```python
 from sklearn.model_selection import train_test_split
 X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=10)
```

```python
X.shape,X_train.shape,X_test.shape
```

```python
#Standardization

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
#Model Training

from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train,y_train)
```

```python
print(regression.coef_)       #8 coeff will be given as their  r8 features
print(regression.intercept_)
```

```python
#Prediction for test data

y_pred = regression.predict(X_test)
y_pred
```

```python
#Performance metrics

from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mse)
sc = r2_score(y_test,y_pred)
print('MSE = ',mse)
print('MAE = ',mae)
print("RMSE = ",rmse)
print("R2-score = ",sc)
```

```python
#Adjusted r2 score

adj_r2 =  1 - ((1-sc)*(len(y_test)-1)/(len(y_test)-X_test.shape[1]-1))
print(adj_r2)
```

```python
plt.scatter(y_test,y_pred)
plt.xlabel("Actual Output")
plt.ylabel("Predicted Output")
plt.title("Actual Output vs Predicted Output")
plt.show()
```

Scatter plot between actual and pred o/p is linear but some pts at end ar not linear so model is not perfectly fit

#Pickling Concept

It is used to pickle ML files into the cloud(deploy) and then it allows to pass any input value from the user side to to the deployed model

```python
#Pickling

import pickle
pickle.dump(regression,open('regression.pkl','wb'))     #saving the regression model in pickle file(serialised format) in write byte mode
```

```python
#loads the pickle file in read byte mode into model
model = pickle.load(open('regression.pkl','rb'))
```
