Independent features = 'tip', 'sex', 'smoker', 'day', 'time',	'size'

Dependent Features = 'total_bill'

```python
#splitting independent and dependent features
X = df.drop('total_bill', axis=1)
y = df['total_bill']
```

```python
#train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=10)
```

Feature Encoding:

1.Label Encoding- this is used to label binary classes as 0/1

2.Onehot Encoding- this is used to label multi-classes in any feature...here 'day' feature

```python
#Feature Encoding
#LabelEncoder is a class
from sklearn.preprocessing import LabelEncoder
le1 = LabelEncoder()
le2 = LabelEncoder()
le3 = LabelEncoder()
```

```python
X_train['sex'] = le1.fit_transform(X_train['sex'])
X_train['smoker'] = le2.fit_transform(X_train['smoker'])
X_train['time'] = le3.fit_transform(X_train['time'])
```

```python
X_train
```

```python
X_test['sex'] = le1.transform(X_test['sex'])
X_test['smoker'] = le2.transform(X_test['smoker'])
X_test['time'] = le3.transform(X_test['time'])
```

```python
#Onehot Encoding---ColumnTransformer

#ColumnTransformer & OneHotEncoder are class
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('onehot', OneHotEncoder(drop='first'), [3])], remainder='passthrough')
```

transformers: This is the main parameter that takes a list of tuples.
Each tuple defines one transformation: (name, transformer_object, columns_to_apply_to)

OneHotEncoder is main object which creates 4 diff binary cols here, as there r 4 categories in 'day' and then drops the first col that it encounters. Coz knowing val of 3 categories model can understand the 4th category val easily, so no need of 1 extra category.

[3] means the 4th col, 'day' needs to be OnehotEncoded

remainder = passthrough lets the other col to remain as it is.

```python
import sys
import numpy
numpy.set_printoptions(threshold=sys.maxsize)

#passing the data on which we need to do transformations
X_train=ct.fit_transform(X_train)
```

```python
X_train
#the encoded binary col for day comes at beginning of this list
```

```python
X_test = ct.transform(X_test)
```

```python
#Apply Support Vector Regression algorithm

from sklearn.svm import SVR
svr = SVR()
```

```python
svr.fit(X_train, y_train)
```

```python
#Performance metrics

y_pred = svr.predict(X_test)
from sklearn.metrics import r2_score,mean_absolute_error
print(r2_score(y_test, y_pred))
print(mean_absolute_error(y_test, y_pred))
```

#HyperParameter Tuning

```python
#gridSearchCV

from sklearn.model_selection import GridSearchCV
param_grid = {'C': [0.1,1,10,100,1000],
              'gamma': [1,0.1,0.01,0.001,0.0001],
              'kernel': ['rbf']}
grid = GridSearchCV(SVR(), param_grid, refit=True, verbose=3)
grid.fit(X_train, y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
grid_pred = grid.predict(X_test)
```

```python
#performance metrics

print(r2_score(y_test, grid_pred))
print(mean_absolute_error(y_test, grid_pred))
```
