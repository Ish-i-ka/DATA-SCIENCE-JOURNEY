```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

Combining the two regions data by adding one extra column named Region

0: Bejaia Region
1: Sidi-Bel Abbes Region



```python
df = dataset
df.loc[:122,"Region"] = 0     #assigning label 0 to first region
df.loc[122:,"Region"] = 1     #assigning label 1 to second region
```

```python
df.info()
```

```python
#Converting data type of Region col from float to int

df[['Region']] = df[['Region']].astype(int)
```

```python
df
```

```python
df.info()
```

```python
#checking for null values

df.isnull().sum()
```

```python
#Removing null values

df = df.dropna().reset_index(drop=True)
```

```python
df.isnull().sum()
```

```python
df.loc[122:123]
```

```python
#so now row 122 shud be removed

df = df.drop(122).reset_index(drop=True)
```

```python
df
```

```python
df.columns
```

```python
#Fix spaces in col names
df.columns = df.columns.str.strip()
df.columns
```

Changing the required columns as int data type #Feature Engineering

```python
df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']] = df[['day', 'month', 'year', 'Temperature', 'RH', 'Ws']].astype(int)
df.info()
```

Changing the remaining col into float

```python
#finds and store the features in 'objects' which are object type only
objects = [features for features in df.columns if df[features].dtypes =='O']
```

```python
for i in objects:
  #type converts every feature into float type except the 'Classes' feature
  if i != 'Classes':
    df[i] =  df[i].astype(float)
```

```python
df.info()
```

```python
df.describe()
```

```python
#Saving the Cleaned Dataset

df.to_csv('Algerian_forest_fires_dataset.csv',index = False)
#index is set to False so that the index col doesnt get appended to the csv
```

###Classification Problem

Dependent Feature = Classes

Independent Feature = Temperature, RH, Ws, Rain, FFMC, DMC, DC, ISI, BUI, FWI

```python
df_copy = df.drop(['day', 'month', 'year'],axis=1)
```

```python
df_copy
```

```python
#Encoding categories in classes

df_copy['Classes'] = np.where(df_copy['Classes'].str.contains('not fire'),0,1)
```

```python
df_copy
```

```python
df_copy['Classes'].value_counts()   #gives the count of different values under the Classes column
```

```python
#Visualization
#Plot Density plot for all features

plt.style.use('seaborn-v0_8-darkgrid')
df_copy.hist(bins=50,figsize=(20,15))
plt.show()
```

```python
#Percent for pie chart

percentage = df_copy['Classes'].value_counts(normalize = True)*100   #normalize=T helps to give the percentage values of the counts of both classes
```

```python
#Plotting pie chart

classlabels = ['Fire','Not Fire']
plt.figure(figsize=(12,7))
plt.pie(percentage,labels=classlabels,autopct='%1.1f%%')        #autopct=auto percentage
plt.title('Pie Chart of Classes')
plt.show()
```
