```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
#Load cancer dataset
from sklearn.datasets import load_breast_cancer
cancer_df = load_breast_cancer()
```

```python
cancer_df.keys()
```

```python
print(cancer_df.DESCR)
```

```python
df = pd.DataFrame(cancer_df['data'],columns = cancer_df['feature_names'])
```

```python
df.head()
```

```python
#Standardization

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(df)
```

```python
scaled_data = scaler.transform(df)
scaled_data
```

Dimensionality Reduction

```python
#Applying PCA algorithm

from sklearn.decomposition import PCA
pca = PCA(n_components=2)     #converting from 30D to 2D
data_pca = pca.fit_transform(scaled_data)
data_pca
```

```python
pca.explained_variance_

#gives the two variance values for the 2 extracted features only
```

```python
cancer_df['target_names']
```

```python
#Visualization

plt.figure(figsize=(8,6))
scatter = plt.scatter(x=data_pca[:,0],y=data_pca[:,1],c=cancer_df['target'],cmap='plasma')    #c adds color according to the class of target feature of the point
plt.xlabel('First Principal Component')
plt.ylabel('Second Principal Component')
plt.title('2D PCA of Breast Cancer Dataset')
handles, auto_labels = scatter.legend_elements()
plt.legend(handles, cancer_df.target_names, title="Diagnosis")
```

```python
#Applying PCA to convert into 3D

pca1 = PCA(n_components=3)
data = pca1.fit_transform(scaled_data)
data
```

```python
pca1.explained_variance_

#PC1,PC2,PC3
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Assuming 'data' contains your 3 principal components from pca1.fit_transform(scaled_data)
scatter=ax.scatter(data[:, 0], data[:, 1], data[:, 2], c=cancer_df['target'], cmap='plasma', s=50)

ax.set_xlabel('First Principal Component')
ax.set_ylabel('Second Principal Component')
ax.set_zlabel('Third Principal Component')
ax.set_title('3D PCA of Breast Cancer Dataset')

handles, labels = scatter.legend_elements()
legend = ax.legend(handles, cancer_df['target_names'], title="Target Classes")

plt.show()
```

Now our two different classes of data points are clearly seperable and we can solve it using classification algorithms
