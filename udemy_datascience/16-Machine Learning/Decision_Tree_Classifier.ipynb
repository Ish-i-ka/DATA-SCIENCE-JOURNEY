```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
print(iris)
```

```python
print(iris['target'])
```

```python
X = pd.DataFrame(iris['data'],columns=iris['feature_names'])
y = iris['target']
```

```python
iris.target_names
```

Here we have to predict the type of flower out of the 3 categories based on independent features : sepal length,sepal width, petal length and petal width

```python
#train test split
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=10)
```

```python
#Apply Decision Tree Classifier
from sklearn.tree import DecisionTreeClassifier
treeclassifier = DecisionTreeClassifier()
treeclassifier.fit(X_train,y_train)
```

```python
#Visualize decision tree

from sklearn import tree
plt.figure(figsize=(15,15))
tree.plot_tree(treeclassifier,filled=True)
```

```python
#Performance Metrics

from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
y_pred = treeclassifier.predict(X_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
```

Now it has been overfit , so we will do Prunning

In Post Pruning , we have to find the ccp_alpha values and then prune the completely formed tree. But this is time taking and usually not preferred

HyperParameter Tuning

```python
#Pre Prunning

params = {
    'criterion':['gini','entropy','log_loss'],
    'splitter':['best','random'],
    'max_depth':[1,2,3,4,5],
    'max_features':['auto','sqrt','log2']
}
```

```python
#removing warnings

import warnings
warnings.filterwarnings('ignore')
```

```python
#GridSearchCV
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(treeclassifier,param_grid=params,cv=5,scoring='accuracy')
grid.fit(X_train,y_train)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
#Performance metrics

from sklearn.metrics import confusion_matrix,classification_report,accuracy_score
y_pred = grid.predict(X_test)
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))
print(accuracy_score(y_test,y_pred))
```
