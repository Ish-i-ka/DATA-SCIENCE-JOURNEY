```python
# closures

def main_welcome():
    msg = "Welcome"
    
    # closure method
    def sub_welcome():
        print(msg)      # the local variable of the outer method can be used inside sub method
        print("Welcome to advanced python")
        print("Please learn these concepts properly")
    return sub_welcome()
```

```python
main_welcome()
```

```python
def main_welcome(func):
    msg = "Welcome"
      
    def sub_welcome():
        func(msg)      # the local variable of the outer method can be used inside sub method
        print("Welcome to advanced python")
        print("Please learn these concepts properly")
    return sub_welcome()

main_welcome(print) # we r passing rint function which replaces func in the method
```

```python
def main_welcome(func,lst):
      
    def sub_welcome():
        print(func(lst))      
        print("Welcome to advanced python")
        print("Please learn these concepts properly")
    return sub_welcome()

main_welcome(len,[1,2,3,4,5])
```

```python
## decorators

def main_welcome(func):
      
    def sub_welcome():
        func()      
        print("Welcome to advanced python")
        print("Please learn these concepts properly")
    return sub_welcome()

def course_intro():
    print("This is an advanced python course.")
    
main_welcome(course_intro)
```

```python
@main_welcome
def course_intro():
    print("This is an advanced python course.")
```

```python
def my_decorator(func):
    def sub_decorator():
        print("You are before the function call")
        func()
        print("You are after the function call")
    return sub_decorator
        
@my_decorator
def say_hello():
    print("Helloooo!!")
```

```python
say_hello()
```

```python
# Decorators with arguments

def repeat(n):
    def decorator(func):
        def wrapper(*args,**kwargs):
            for i in range(n):
                func(*args,**kwargs)
        return wrapper
    return decorator
```

```python
@repeat(3)
def say_hello():
    print("Hello")
```

```python
say_hello()
```
